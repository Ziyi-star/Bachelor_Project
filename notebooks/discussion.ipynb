```python
import numpy as np
import matplotlib.pyplot as plt
```

```python


def plot_acceleration_example():
    # Create time array (0 to 3 seconds with 1000 points)
    t = np.linspace(0, 3, 1000)
    
    np.random.seed(42)  # For reproducibility
    noise = np.random.normal(0, 0.1, len(t))
    base = 0.2 * np.sin(20 * t) + 0.1 + noise
    
    # Create two peaks
    def gaussian_peak(x, mu, sigma, amplitude):
        return amplitude * np.exp(-(x - mu)**2 / (2 * sigma**2))
    
    # Add two Gaussian peaks
    peak1 = gaussian_peak(t, 1.55, 0.01, 5)  # Increased amplitude to 5
    peak2 = gaussian_peak(t, 1.85, 0.01, 5)  # Increased amplitude to 5
    
    # Combine signals
    signal = base + peak1 + peak2
    
    # Create the plot
    plt.figure(figsize=(10, 6))
    
    # Plot the main signal
    plt.plot(t, signal, 'b-', linewidth=2)
    
    # Add vertical lines
    plt.axvline(x=1.0, color='r', linestyle='-', alpha=0.5)
    plt.axvline(x=2.0, color='r', linestyle='-', alpha=0.5)
    
    # # Add text annotations
    # plt.text(0.5, -0.5, '0.5 s', color='green', ha='center')
    # plt.text(1.5, -0.5, '0.5 s', color='green', ha='center')
    
    # Customize the plot
    plt.xlabel('Time (s)', fontsize=14)
    plt.ylabel('Acceleration on Z axis', fontsize=14)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.grid(True, alpha=0.3)
    
    # Set axis limits
    plt.ylim(-1, 5)
    plt.xlim(0, 3)
    
    # Remove top and right spines
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)
    
    plt.tight_layout()
    plt.show()

```

```python
# Call the function to create the plot
plot_acceleration_example()
```
